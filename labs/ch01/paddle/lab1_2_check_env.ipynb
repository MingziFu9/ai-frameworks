### 第 4 步：让框架做一道大题，计时

```python
a = paddle.rand([2000, 2000])                       # 2000×2000 的随机数表格
start = time.time()
c = paddle.matmul(a, a)                             # 矩阵乘法：约 80 亿次乘法和加法
print(f"2000×2000 矩阵乘法用时：{time.time() - start:.3f} 秒")
```
